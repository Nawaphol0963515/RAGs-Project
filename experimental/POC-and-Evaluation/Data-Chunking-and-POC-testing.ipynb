```python
import os
import glob
import getpass
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.prompts import PromptTemplate # <--- เพิ่ม Import ตัวนี้สำหรับทำ Metadata Prompting


# ==========================================
# Config API key
if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter Gemini api key: ")

data_directory = "/content/drive/MyDrive/RAGs-Raw-Data"  # Root folder containing 5 candidates KBANK, BBL, KTB, SCBN and PTT of 56-1 Structure report files
# ==========================================

# ==========================================
# 1. INGESTION & CHUNKING (MULTI-DOCUMENT)
print("1. Loading all PDFs from directory")
all_docs = []

# load all .pdf in the root folder
pdf_files = glob.glob(os.path.join(data_directory, "*.PDF"))

if not pdf_files:
    print(f"Error: No PDF files found in {data_directory}")
else:
    for file_path in pdf_files:
        print(f" -> Reading: {os.path.basename(file_path)}")
        loader = PyPDFLoader(file_path)
        # PyPDFLoader use Metadata Tag ==> {"source": "filename.pdf"} Auto by the pypdf library
        all_docs.extend(loader.load())

print(f"\nTotal pages loaded across all companies: {len(all_docs)}")

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=["\n\n", "\n", " ", ""]
)

chunks = text_splitter.split_documents(all_docs)
print(f"-> Created {len(chunks)} total chunks across all documents.")
# ==========================================

# ==========================================
# 2. EMBEDDING & LOCAL VECTOR STORE
print("\n2. Initializing Multilingual Embeddings and ChromaDB...")

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)  # use this one because its the most support for Thai language

vector_store = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="./multi_company_chroma_db"
)

# Change k value to be 15 or 20 for selecting the chunk to cover all of the 5 company (more than 5)
# in this case let use MMR to reduce duplicate chunks
retriever = vector_store.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 8, "fetch_k": 20} # adjust this one
)
print("-> Vectors successfully saved to local ChromaDB")
# ==========================================

# ==========================================
# 3. LLM setup with the system prompt
print("\n3. Setting up Gemini LLM...")

llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview", temperature=0)

system_prompt = (
    "You are an expert Thai Financial Analyst. "
    "Use the following pieces of retrieved financial context to answer the question. "
    "The context contains data from multiple companies. Make sure to clearly identify which data belongs to which company based on the document source or context. "
    "If the context does not contain the answer for a specific company, explicitly state that the data is not available for that company. "
    "Answer in clear, professional Thai language, and use bullet points or a table format if comparing multiple companies."
    "\n\n"
    "Context: {context}"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])

# ---> edit: สร้าง Template บังคับให้พิมพ์ชื่อไฟล์ แปะไว้บนหัวข้อความทุกก้อน
document_prompt = PromptTemplate.from_template(
    "ชื่อไฟล์เอกสาร (Source): {source}\nเนื้อหา: {page_content}"
)

# ---> edit: นำ document_prompt เข้าไปประกอบใน chain เพื่อให้ AI เห็น Metadata
question_answer_chain = create_stuff_documents_chain(
    llm=llm,
    prompt=prompt,
    document_prompt=document_prompt
)
rag_chain = create_retrieval_chain(retriever, question_answer_chain)
# ==========================================

# ==========================================
# 4. Cross-document inference testing with user prompt
print("\n4. Running Multi-Document RAG Inference Test...")

question = "รายได้ดอกเบี้ยสุทธิของธนาคาร KBANK, KTB, BBL, SCB เเละ PTT ในปี 2567 และ 2568 คือเท่าไร?"

print(f"\nQuestion: {question}")
print("Thinking (need to be across all document)\n")

# ------------------------------------------
# Show retrieved chunks with source + page
retrieved_docs = retriever.invoke(question)

print("===== RETRIEVED CHUNKS (SOURCE / PAGE) =====")
for i, doc in enumerate(retrieved_docs, 1):
    source_name = os.path.basename(doc.metadata.get("source", "unknown"))
    page_no = doc.metadata.get("page", "unknown")
    preview = doc.page_content[:300].replace("\n", " ")

    print(f"\nChunk #{i}")
    print(f"Source: {source_name}")
    print(f"Page: {page_no}")
    print(f"Preview: {preview}...")
print("============================================\n")
# ------------------------------------------

response = rag_chain.invoke({"input": question})

print("================ ANSWER ================")
print(response["answer"])
print("========================================")
# ==========================================
```

## **About Evalution Strategy**
Due to the project objective and some of the situaiton in the timeline, full automated evaluation (Ragas or TruLens) was out of scope. Instead, I engineered a Golden Dataset of 15 edge-case financial queries across 5 institutions to manually verify Retrieval Precision and Generation Faithfulness.
Sine our objective is to build a demonstration to study a RAGs sysytem and how really we can integrate to the apllication so I have decided to use this one as a first phase for RAGs implementation and testing since I still have to work on the project in my Master study also so in this first phase we will do roughly in evaluation and then try to deminstrate the applicaiton in use as a chatbot ui. If there is more convinient time I will surely continue with the full pipeline to testing the RAGs with my idea to try for the Thai pdf in financial of the SET from 56-1 report which I did not found some sourecs of data yet and its required much time time to setup everything

FYI:
- How RAGs actually does need in evaluation.
In a real enterprise, you cannot just test 5 questions yourself and deploy it instead we need to prove mathematically that the our AI and Retrieval information from RAGs is not hallucinating.
We can use automated evaluation frameworks (like Ragas or TruLens) to score RAGs systems on two main things:

1. Retrieval Metrics (Is the Database working?):
 * Context Precision: Did ChromaDB fetch the exact right chunk, or did it pull in useless junk?
 * Context Recall: Did ChromaDB miss any important chunks?

2. Generation Metrics (Is the LLM working?):
 * Faithfulness: Did the AI make up a number, or is the answer 100% grounded in the retrieved chunks?
 * Answer Relevance: Did the AI actually answer the user's prompt, or did it go off-topic?

```python
import os
import glob
import getpass
import time
import shutil
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# ==========================================
# config and setup
if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter Gemini api key: ")

data_directory = "/content/drive/MyDrive/RAGs-Raw-Data"

print("Loading PDFs once to save time...")
all_docs = []
pdf_files = glob.glob(os.path.join(data_directory, "*.PDF"))
for file_path in pdf_files:
    loader = PyPDFLoader(file_path)
    all_docs.extend(loader.load())

# Initialize embedding model (THai model)
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

# Representative Questions for the Grid Search
mini_dataset = [
    {"type": "Single", "q": "รายได้ดอกเบี้ยสุทธิของธนาคาร KBANK ในปี 2567 คือเท่าไร?", "ans": "148,004,290.00 พันบาท"},
    {"type": "Negative", "q": "รายได้ดอกเบี้ยสุทธิของบริษัท PTT ในปี 2567 คือเท่าไร?", "ans": "ไม่มีข้อมูล / PTT ไม่ใช่ธุรกิจธนาคาร"},
    {"type": "Cross", "q": "เปรียบเทียบรายได้ดอกเบี้ยสุทธิของธนาคาร KBANK และ BBL ในปี 2567", "ans": "KBANK = 148,004,290.00 พันบาท, BBL = 133,899,581.00 พันบาท"}
]

# Set up the LLM as a judge
evaluator_llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview", temperature=0)

eval_prompt = PromptTemplate.from_template(
    """You are an expert AI Engineer evaluating a RAG system.
    The input data will be in Thai. Your job is to compare the actual answer against the expected fact.
    Q: {question}
    Expected: {expected_fact}
    Actual: {actual_answer}
    Task: Evaluate Thai RAG factual accuracy. Determine if the 'Actual Answer' contains the same factual information as the 'Expected Fact'.
    Ignore phrasing; focus STRICTLY on numbers, names, and core facts. ('No data' = 'Cannot be found' -> PASS).
    Output EXACTLY in this format:
    RESULT: [PASS or FAIL] [Score]/100
    REASON: [1 short Thai sentence explaining the score and result]"""
)
eval_chain = eval_prompt | evaluator_llm | StrOutputParser()

# ==========================================
# 1. The evaluation Function
def test_rag_setting(setting_name, chunk_size, chunk_overlap, search_type, k_val, fetch_k_val=None):
    print(f"\n{'='*50}")
    print(f" RUNNING TEST: {setting_name}")
    print(f"Parameters: Chunk={chunk_size}, Overlap={chunk_overlap}, Search={search_type}, k={k_val}")
    print(f"{'='*50}")

    # 1. Chunking
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size, chunk_overlap=chunk_overlap, separators=["\n\n", "\n", " ", ""]
    )
    chunks = text_splitter.split_documents(all_docs)

    # 2. Temporary Vector Store (Stored in memory for fast testing)(ChromaDB)
    vector_store = Chroma.from_documents(documents=chunks, embedding=embeddings)

    # 3. Retriever Settings
    if search_type == "mmr":
        retriever = vector_store.as_retriever(search_type="mmr", search_kwargs={"k": k_val, "fetch_k": fetch_k_val})
    else:
        retriever = vector_store.as_retriever(search_kwargs={"k": k_val})

    # 4. RAG Chain Setup
    llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview", temperature=0)
    system_prompt = "You are an expert Thai Financial Analyst. Context: {context}"
    prompt = ChatPromptTemplate.from_messages([("system", system_prompt), ("human", "{input}")])
    document_prompt = PromptTemplate.from_template("Source: {source}\nContent: {page_content}")

    question_answer_chain = create_stuff_documents_chain(llm, prompt, document_prompt=document_prompt)
    rag_chain = create_retrieval_chain(retriever, question_answer_chain)

    # 5. Evaluate the representative quesiton
    passed = 0
    for idx, test in enumerate(mini_dataset, 1):
        print(f"  -> Testing Q{idx}: {test['type']} ... ", end="", flush=True)

        # for free tier api
        time.sleep(4)

        try:
            rag_response = rag_chain.invoke({"input": test['q']})
            eval_result = eval_chain.invoke({
                "question": test['q'], "expected_fact": test['ans'], "actual_answer": rag_response["answer"]
            })

            if "RESULT: PASS" in eval_result.upper():
                passed += 1
                print("1: PASS")
            else:
                print("0: FAIL")
        except Exception as e:
            print(f"!!!ERROR (Likly to be API Limit): {e}")

    # Clean up the database from memory to prepare for the next test
    vector_store.delete_collection()

    score = (passed / len(mini_dataset)) * 100
    print(f"Score for {setting_name}: {score:.1f}%")
    return score

# ==========================================
# 2. RUN THE GRID SEARCH
# ==========================================
print("\nStarting Hyperparameter Grid Search...")

# SETTING 1: Basic & Small (Standard default, usually fails Cross-Document)
score_1 = test_rag_setting("Setting 1 (Basic)", chunk_size=500, chunk_overlap=50, search_type="similarity", k_val=5)

# SETTING 2: Medium MMR (The balanced enterprise approach)
score_2 = test_rag_setting("Setting 2 (Medium MMR)", chunk_size=1000, chunk_overlap=200, search_type="mmr", k_val=10, fetch_k_val=30)

# SETTING 3: Large MMR (Heavy context, high k value)
score_3 = test_rag_setting("Setting 3 (Large MMR)", chunk_size=2000, chunk_overlap=400, search_type="mmr", k_val=20, fetch_k_val=50)

# ==========================================
# 3. CONCLUSION
# ==========================================
print("\n" + "="*50)
print("GRID SEARCH RESULTS")
print(f"Setting 1: {score_1}%")
print(f"Setting 2: {score_2}%")
print(f"Setting 3: {score_3}%")
print("="*50)
```

```python
import os
import glob
import getpass
import time
import re
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# =====================================================================
# CHECKPOINT DATA: From your previous run
# Tests 1-10 passed, Test 11-12 failed, Test 13 passed. Total = 11
PREVIOUS_PASSES = 11
TOTAL_QUESTIONS = 15
# =====================================================================

# CONFIGURATION: Using your winning Setting 2 parameters
BEST_CHUNK_SIZE = 1000
BEST_CHUNK_OVERLAP = 200
BEST_SEARCH_TYPE = "mmr"
BEST_K_VAL = 10
BEST_FETCH_K_VAL = 30

# 0. API KEY SETUP (Enter your NEW key here)
os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your NEW Gemini api key: ")

data_directory = "/content/drive/MyDrive/RAGs-Raw-Data"

# ==========================================
# 1. INGESTION & CHUNKING
# ==========================================
print(f"\n--- 1. Rebuilding Database for Final Checkpoint ---")
all_docs = []
pdf_files = glob.glob(os.path.join(data_directory, "*.PDF"))
for file_path in pdf_files:
    loader = PyPDFLoader(file_path)
    all_docs.extend(loader.load())

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=BEST_CHUNK_SIZE,
    chunk_overlap=BEST_CHUNK_OVERLAP,
    separators=["\n\n", "\n", " "]
)
chunks = text_splitter.split_documents(all_docs)

# ==========================================
# 2. EMBEDDING & RETRIEVER
# ==========================================
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
vector_store = Chroma.from_documents(documents=chunks, embedding=embeddings)

retriever = vector_store.as_retriever(
    search_type="mmr",
    search_kwargs={"k": BEST_K_VAL, "fetch_k": BEST_FETCH_K_VAL}
)

# ==========================================
# 3. RAG CHAIN & JUDGE SETUP
# ==========================================
# Using 1.5-flash for the Judge to ensure high daily quota availability
rag_llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview", temperature=0)
judge_llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0)

system_prompt = (
    "You are an expert Thai Financial Analyst. Use the following retrieved financial context to answer. "
    "Identify which data belongs to which company. If data is missing, say it's not available. "
    "\n\nContext: {context}"
)
prompt = ChatPromptTemplate.from_messages([("system", system_prompt), ("human", "{input}")])
document_prompt = PromptTemplate.from_template("ชื่อไฟล์เอกสาร (Source): {source}\nเนื้อหา: {page_content}")

question_answer_chain = create_stuff_documents_chain(rag_llm, prompt, document_prompt=document_prompt)
rag_chain = create_retrieval_chain(retriever, question_answer_chain)

eval_prompt = PromptTemplate.from_template(
    """Evaluate Thai RAG factual accuracy. Focus STRICTLY on numbers, names, and core facts.
    Q: {question}
    Expected: {expected_fact}
    Actual: {actual_answer}
    Output EXACTLY:
    RESULT: [PASS or FAIL] [Score]/100
    REASON: [1 short Thai sentence]"""
)
eval_chain = eval_prompt | judge_llm | StrOutputParser()

# ==========================================
# HELPER: ROBUST JUDGE OUTPUT PARSER
# ==========================================
def parse_judge_output(eval_result_text):
    text = str(eval_result_text).strip()

    # Detect PASS / FAIL even if model returns:
    # RESULT: PASS 100/100
    # RESULT: [PASS] 100/100
    # RESULT:[FAIL] 45/100
    result_match = re.search(r"RESULT\s*:\s*\[?\s*(PASS|FAIL)\s*\]?", text, flags=re.IGNORECASE)
    result = result_match.group(1).upper() if result_match else None

    # Extract score like 100/100 or 50/100
    score_match = re.search(r"(\d{1,3}\s*/\s*100)", text, flags=re.IGNORECASE)
    score = score_match.group(1) if score_match else None

    # Extract reason
    reason_match = re.search(r"REASON\s*:\s*(.*)", text, flags=re.IGNORECASE | re.DOTALL)
    reason = reason_match.group(1).strip() if reason_match else text

    return result, score, reason

# ==========================================
# 4. REMAINING EVALUATION (Questions 14 & 15)
# ==========================================
remaining_dataset = [
    {"idx": 14, "type": "Cross", "q": "จากเอกสารทั้งหมด มีบริษัทใดบ้างที่ระบุว่าประกอบธุรกิจ 'ธนาคารพาณิชย์'", "ans": "KBANK, BBL, KTB, SCB"},
    {"idx": 15, "type": "Cross", "q": "รายงานของธนาคาร KBANK, SCB และ BBL เป็นข้อมูลสิ้นสุด ณ วันที่และปีใดเหมือนกัน?", "ans": "31 ธันวาคม 2568"}
]

print("\n--- 4. Evaluating Remaining Questions (14 & 15) ---")
new_passes = 0

for test in remaining_dataset:
    print(f"[Test {test['idx']}/15 - {test['type']}] Q: {test['q']}")
    try:
        rag_response = rag_chain.invoke({"input": test['q']})
        eval_result = eval_chain.invoke({
            "question": test['q'],
            "expected_fact": test['ans'],
            "actual_answer": rag_response["answer"]
        })

        parsed_result, parsed_score, parsed_reason = parse_judge_output(eval_result)

        if parsed_result == "PASS":
            new_passes += 1
            print("RESULT: 1 --> PASS")
        else:
            print("RESULT: 0 --> FAIL")

        if parsed_score:
            print(f"Judge Reason: {parsed_score}")
            print(f"REASON: {parsed_reason}\n")
        else:
            print(f"Judge Reason: {parsed_reason}\n")

    except Exception as e:
        print(f"ERROR on Test {test['idx']}: {e}\n")

    time.sleep(8)  # Safety delay

# ==========================================
# 5. FINAL COMBINED RESULT
# ==========================================
final_passed = PREVIOUS_PASSES + new_passes
final_accuracy = (final_passed / TOTAL_QUESTIONS) * 100

print("=" * 50)
print("FINAL COMBINED RAG EVALUATION REPORT")
print("=" * 50)
print(f"Previously Passed:  {PREVIOUS_PASSES} / 13")
print(f"Newly Passed:       {new_passes} / 2")
print("-" * 50)
print(f"TOTAL ACCURACY:     {final_passed} / {TOTAL_QUESTIONS} ({final_accuracy:.1f}%)")
print("=" * 50)

# Clean up
vector_store.delete_collection()
```
